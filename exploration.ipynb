```python
import pyopenms as oms
import pandas as pd
import numpy as np 
import scipy 
import os 
import glob
```

```python
samples = glob.glob("data/mzml/*")
samples
```

```python
exp = oms.MSExperiment()
oms.MzMLFile().load(samples[0], exp)
exp.sortSpectra(True)
```

```python
mtd = oms.MassTraceDetection()
params = mtd.getDefaults()
params.setValue("mass_error_ppm", 5.0)    
params.setValue("noise_threshold_int", 1000.0)
mtd.setParameters(params)
mass_traces = []
mtd.run(exp,mass_traces,0)
```

```python
print(f"Detected {len(mass_traces)} mass traces")
```
